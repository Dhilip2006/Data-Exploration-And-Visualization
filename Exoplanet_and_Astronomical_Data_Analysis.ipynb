Exoplanet and Astronomical Data Analysis

This project analyzes exoplanet data from the NASA Exoplanet Archive using Python. It applies concepts from all five units of Data Exploration and Visualization, including data cleaning, exploratory data analysis, statistical analysis, data visualization, correlation analysis, PCA, and machine learning. The objective is to identify meaningful patterns in exoplanet characteristics and discovery trends through effective data analysis and visualization.

```python
import os
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from scipy import stats
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.decomposition import PCA
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score, classification_report,
    ConfusionMatrixDisplay
)

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)

# LOAD DATASET

file_path = "Exoplanet and Astronomical Data.csv"

if not os.path.exists(file_path):
    raise FileNotFoundError(
        "Place the NASA CSV file in the same folder as this program."
    )

# NASA CSV files contain metadata comment lines beginning with #
df = pd.read_csv(file_path, comment="#", low_memory=False)

print("=" * 65)
print("EXOPLANET AND ASTRONOMICAL DATA ANALYSIS")
print("=" * 65)

print("\nDataset shape:", df.shape)
print("\nColumn names:\n", df.columns.tolist())
print("\nFirst five records:\n", df.head())
print("\nDataset information:")
df.info()

# EXPLORATORY DATA ANALYSIS

print("\n\nEXPLORATORY DATA ANALYSIS")

# Dataset dimensions and structure
print("\nNumber of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

# Missing values
missing = df.isnull().sum()
missing = missing[missing > 0].sort_values(ascending=False)

print("\nColumns with missing values:\n", missing)

# Missing-value percentage
missing_percentage = (df.isnull().mean() * 100).round(2)
print("\nMissing-value percentage:\n",
      missing_percentage[missing_percentage > 0])

# Visualize missing values for the most incomplete columns
if len(missing) > 0:
    missing.head(15).sort_values().plot(
        kind="barh", figsize=(10, 6), title="Top Columns by Missing Values"
    )
    plt.xlabel("Number of missing values")
    plt.tight_layout()
    plt.show()

# Duplicate records
print("\nCompletely duplicated rows:", df.duplicated().sum())
print("Unique planet names:", df["pl_name"].nunique())
analysis_df = df.drop_duplicates().copy()

# Convert numeric columns safely
numeric_cols = [
    "sy_snum", "sy_pnum", "disc_year", "pl_orbper",
    "pl_orbsmax", "pl_rade", "pl_masse", "pl_dens",
    "pl_orbeccen", "pl_eqt", "st_teff", "st_rad",
    "st_mass", "st_met"
]

for col in numeric_cols:
    if col in analysis_df.columns:
        analysis_df[col] = pd.to_numeric(
            analysis_df[col], errors="coerce"
        )

# Select useful columns
selected_cols = [
    "pl_name", "hostname", "sy_snum", "sy_pnum",
    "discoverymethod", "disc_year", "disc_facility",
    "pl_orbper", "pl_orbsmax", "pl_rade", "pl_masse",
    "pl_dens", "pl_orbeccen", "pl_eqt", "st_teff",
    "st_rad", "st_mass", "st_met"
]

selected_cols = [
    c for c in selected_cols if c in analysis_df.columns
]

planet_df = analysis_df[selected_cols].copy()

# Slicing and filtering
print("\nPlanets discovered after 2015:")
print(
    planet_df.loc[
        planet_df["disc_year"] > 2015,
        ["pl_name", "disc_year", "discoverymethod"]
    ].head(10)
)

# Grouping and aggregation
print("\nDiscovery-method counts:")
print(planet_df["discoverymethod"].value_counts().head(10))

print("\nAverage planetary measurements:")
print(
    planet_df.groupby("discoverymethod")[
        [c for c in ["pl_rade", "pl_masse", "pl_eqt"]
         if c in planet_df.columns]
    ].mean(numeric_only=True).round(2)
)

planet_df.to_csv("exoplanet_analysis_data.csv", index=False)
print("\nCleaned analysis dataset saved.")

# VISUALIZATION USING MATPLOTLIB

print("\n\nMATPLOTLIB VISUALIZATION")

# Line plot: annual discovery trend
yearly = (
    planet_df.dropna(subset=["disc_year"])
    .groupby("disc_year")["pl_name"]
    .nunique()
    .sort_index()
)

plt.figure()
plt.plot(yearly.index, yearly.values, marker="o")
plt.title("Unique Exoplanets by Discovery Year")
plt.xlabel("Discovery year")
plt.ylabel("Number of unique planets")
plt.grid(True)
plt.tight_layout()
plt.show()

# Histogram: planetary radius
radius = planet_df["pl_rade"].dropna()
radius = radius[radius > 0]

plt.figure()
plt.hist(radius, bins=40, edgecolor="black")
plt.title("Distribution of Exoplanet Radii")
plt.xlabel("Planet radius (Earth radii)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

# Scatter plot: orbital period vs radius
scatter_df = planet_df[
    ["pl_orbper", "pl_rade"]
].dropna()

scatter_df = scatter_df[
    (scatter_df["pl_orbper"] > 0) &
    (scatter_df["pl_rade"] > 0)
]

plt.figure()
plt.scatter(
    scatter_df["pl_orbper"],
    scatter_df["pl_rade"],
    alpha=0.5
)
plt.xscale("log")
plt.yscale("log")
plt.title("Orbital Period vs Planet Radius")
plt.xlabel("Orbital period (days, log scale)")
plt.ylabel("Planet radius (Earth radii, log scale)")
plt.tight_layout()
plt.show()

# Error visualization using measurement uncertainties
# Upper and lower uncertainty fields are supplied in the NASA file.
error_cols = [
    "pl_rade", "pl_radeerr1", "pl_radeerr2"
]

if all(c in analysis_df.columns for c in error_cols):
    err_df = analysis_df[error_cols].dropna().copy()
    err_df = err_df[
        (err_df["pl_rade"] > 0) &
        (err_df["pl_radeerr1"] >= 0) &
        (err_df["pl_radeerr2"] <= 0)
    ].head(25)

    if not err_df.empty:
        plt.figure(figsize=(12, 6))
        plt.errorbar(
            np.arange(len(err_df)),
            err_df["pl_rade"],
            yerr=[
                err_df["pl_radeerr1"],
                -err_df["pl_radeerr2"]
            ],
            fmt="o",
            capsize=3
        )
        plt.title("Planet Radius with Measurement Uncertainty")
        plt.xlabel("Planet record")
        plt.ylabel("Planet radius (Earth radii)")
        plt.tight_layout()
        plt.show()

# Subplots
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

axes[0].hist(radius, bins=30, edgecolor="black")
axes[0].set_title("Planet Radius")
axes[0].set_xlabel("Earth radii")

period = planet_df["pl_orbper"].dropna()
period = period[period > 0]

axes[1].hist(np.log10(period), bins=30, edgecolor="black")
axes[1].set_title("Log Orbital Period")
axes[1].set_xlabel("log10(days)")

plt.tight_layout()
plt.show()

# Density curve
if len(radius) > 1:
    sns.kdeplot(x=radius, fill=True)
    plt.title("Density Curve of Planet Radius")
    plt.xlabel("Planet radius (Earth radii)")
    plt.tight_layout()
    plt.show()

# 3D scatter plot
from mpl_toolkits.mplot3d import Axes3D

three_cols = ["pl_rade", "pl_eqt", "st_teff"]
if all(c in planet_df.columns for c in three_cols):
    three_df = planet_df[three_cols].dropna()
    three_df = three_df[
        (three_df["pl_rade"] > 0) &
        (three_df["pl_eqt"] > 0) &
        (three_df["st_teff"] > 0)
    ].head(2000)

    if len(three_df) > 0:
        fig = plt.figure()
        ax = fig.add_subplot(111, projection="3d")
        ax.scatter(
            three_df["pl_rade"],
            three_df["pl_eqt"],
            three_df["st_teff"],
            alpha=0.5
        )
        ax.set_xlabel("Planet radius")
        ax.set_ylabel("Equilibrium temperature")
        ax.set_zlabel("Stellar temperature")
        ax.set_title("3D Exoplanet Properties")
        plt.tight_layout()
        plt.show()

# UNIVARIATE ANALYSIS
print("\n\nUNIVARIATE ANALYSIS")

# Descriptive statistics
measure_cols = [
    c for c in [
        "pl_orbper", "pl_rade", "pl_masse", "pl_dens",
        "pl_eqt", "pl_orbeccen", "st_teff", "st_mass"
    ]
    if c in planet_df.columns
]

print("\nDescriptive statistics:")
print(planet_df[measure_cols].describe().T)

# Mean, median, standard deviation, skewness
for col in measure_cols:
    values = planet_df[col].dropna()

    if len(values) > 0:
        print(f"\n{col}")
        print("Mean:", values.mean())
        print("Median:", values.median())
        print("Standard deviation:", values.std())
        print("Skewness:", values.skew())

# Box plots for outlier visualization
box_cols = [
    c for c in ["pl_rade", "pl_eqt", "st_teff"]
    if c in planet_df.columns
]

fig, axes = plt.subplots(1, len(box_cols), figsize=(5 * len(box_cols), 5))

if len(box_cols) == 1:
    axes = [axes]

for ax, col in zip(axes, box_cols):
    values = planet_df[col].dropna()
    ax.boxplot(values, vert=True)
    ax.set_title(f"Outlier Analysis: {col}")
    ax.set_ylabel(col)

plt.tight_layout()
plt.show()

# IQR-based outlier counts
for col in measure_cols:
    values = planet_df[col].dropna()

    if len(values) > 0:
        q1 = values.quantile(0.25)
        q3 = values.quantile(0.75)
        iqr = q3 - q1

        outliers = values[
            (values < q1 - 1.5 * iqr) |
            (values > q3 + 1.5 * iqr)
        ]

        print(f"{col}: {len(outliers)} IQR outliers")


# Frequency distribution of discovery methods
method_counts = planet_df["discoverymethod"].value_counts()

method_counts.plot(kind="bar", figsize=(12, 6))
plt.title("Frequency of Exoplanet Discovery Methods")
plt.xlabel("Discovery method")
plt.ylabel("Number of records")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

# Pie chart of the most common methods
top_methods = method_counts.head(6)

plt.figure()
plt.pie(
    top_methods.values,
    labels=top_methods.index,
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Most Common Discovery Methods")
plt.tight_layout()
plt.show()

# Frequency table
print("\nDiscovery-method frequency table:")
print(pd.DataFrame({
    "Frequency": method_counts,
    "Percentage": (
        method_counts / method_counts.sum() * 100
    ).round(2)
}))

# BIVARIATE ANALYSIS
print("\n\nBIVARIATE ANALYSIS")

# Pearson and Spearman correlation
pair_df = planet_df[
    ["pl_rade", "pl_eqt", "st_teff", "pl_orbper", "st_mass"]
].copy()

# Use only positive values for log-scaled physical variables
for col in ["pl_rade", "pl_eqt", "st_teff", "pl_orbper", "st_mass"]:
    pair_df.loc[pair_df[col] <= 0, col] = np.nan

print("\nPearson correlation:")
print(pair_df.corr(method="pearson").round(3))

print("\nSpearman correlation:")
print(pair_df.corr(method="spearman").round(3))

# Scatter plot: planet radius vs equilibrium temperature
biv = planet_df[["pl_rade", "pl_eqt"]].dropna()
biv = biv[(biv["pl_rade"] > 0) & (biv["pl_eqt"] > 0)]

plt.figure()
plt.scatter(biv["pl_rade"], biv["pl_eqt"], alpha=0.5)
plt.xscale("log")
plt.title("Planet Radius vs Equilibrium Temperature")
plt.xlabel("Planet radius (Earth radii, log scale)")
plt.ylabel("Equilibrium temperature (K)")
plt.tight_layout()
plt.show()

# Correlation test
test_df = planet_df[["pl_rade", "pl_eqt"]].dropna()
test_df = test_df[
    (test_df["pl_rade"] > 0) & (test_df["pl_eqt"] > 0)
]

if len(test_df) >= 3:
    pearson_r, pearson_p = stats.pearsonr(
        test_df["pl_rade"], test_df["pl_eqt"]
    )
    spearman_r, spearman_p = stats.spearmanr(
        test_df["pl_rade"], test_df["pl_eqt"]
    )

    print("\nRadius vs temperature correlation")
    print("Pearson r:", pearson_r, "p-value:", pearson_p)
    print("Spearman rho:", spearman_r, "p-value:", spearman_p)

# Box plot: planet radius by discovery method
top_method_names = method_counts.head(6).index
group_df = planet_df[
    planet_df["discoverymethod"].isin(top_method_names)
].copy()

sns.boxplot(
    data=group_df,
    x="discoverymethod",
    y="pl_rade"
)
plt.yscale("log")
plt.title("Planet Radius by Discovery Method")
plt.xlabel("Discovery method")
plt.ylabel("Planet radius (Earth radii, log scale)")
plt.xticks(rotation=35, ha="right")
plt.tight_layout()
plt.show()

# Violin plot
sns.violinplot(
    data=group_df,
    x="discoverymethod",
    y="pl_eqt",
    cut=0
)
plt.title("Temperature Distribution by Discovery Method")
plt.xlabel("Discovery method")
plt.ylabel("Equilibrium temperature (K)")
plt.xticks(rotation=35, ha="right")
plt.tight_layout()
plt.show()

# Grouped bar chart: discoveries by decade
decade_df = planet_df.dropna(subset=["disc_year"]).copy()
decade_df["decade"] = (
    decade_df["disc_year"].astype(int) // 10 * 10
)

decade_method = pd.crosstab(
    decade_df["decade"],
    decade_df["discoverymethod"]
)

top_for_bar = method_counts.head(5).index
decade_method[top_for_bar].plot(
    kind="bar", figsize=(13, 6)
)
plt.title("Discovery Methods Across Decades")
plt.xlabel("Discovery decade")
plt.ylabel("Number of records")
plt.tight_layout()
plt.show()

# UNIT 5: MULTIVARIATE ANALYSIS
print("\n\nMULTIVARIATE ANALYSIS")

# Pair plot
pair_plot_cols = ["pl_rade", "pl_eqt", "st_teff", "pl_orbper"]
pair_plot_df = planet_df[pair_plot_cols].copy()

for col in pair_plot_cols:
    pair_plot_df.loc[pair_plot_df[col] <= 0, col] = np.nan

pair_plot_df = pair_plot_df.dropna().sample(
    n=min(500, len(pair_plot_df)), random_state=42
)

if len(pair_plot_df) > 1:
    sns.pairplot(pair_plot_df, corner=True)
    plt.suptitle("Pairwise Relationships Between Exoplanet Features", y=1.02)
    plt.show()

# Correlation heatmap
heatmap_df = pair_df.dropna(how="all")
plt.figure(figsize=(10, 7))
sns.heatmap(
    heatmap_df.corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    center=0
)
plt.title("Correlation Heatmap of Astronomical Properties")
plt.tight_layout()
plt.show()

# Prepare numerical features for PCA
pca_features = [
    c for c in [
        "pl_rade", "pl_eqt", "st_teff",
        "st_mass", "st_rad", "pl_orbper",
        "pl_orbsmax", "pl_orbeccen", "st_met"
    ]
    if c in planet_df.columns
]

pca_data = planet_df[pca_features].copy()

# Replace nonphysical zero or negative values where appropriate
for col in [
    "pl_rade", "pl_eqt", "st_teff", "st_mass",
    "st_rad", "pl_orbper", "pl_orbsmax"
]:
    if col in pca_data.columns:
        pca_data.loc[pca_data[col] <= 0, col] = np.nan

# Remove entirely empty features
pca_data = pca_data.dropna(axis=1, how="all")

if pca_data.shape[1] >= 2 and len(pca_data) >= 3:
    # Impute missing values for PCA, then standardize
    imputer = SimpleImputer(strategy="median")
    X_imputed = imputer.fit_transform(pca_data)

    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X_imputed)

    pca = PCA(n_components=2)
    components = pca.fit_transform(X_scaled)

    pca_df = pd.DataFrame(
        components,
        columns=["PC1", "PC2"]
    )

    print("\nPCA explained variance ratio:",
          pca.explained_variance_ratio_)
    print("Total variance explained by two components:",
          pca.explained_variance_ratio_.sum())

    plt.figure()
    plt.scatter(
        pca_df["PC1"], pca_df["PC2"], alpha=0.5
    )
    plt.title("PCA of Exoplanet and Stellar Properties")
    plt.xlabel("Principal Component 1")
    plt.ylabel("Principal Component 2")
    plt.tight_layout()
    plt.show()

# Machine learning: classify discovery method

ml_features = [
    c for c in [
        "pl_rade", "pl_eqt", "st_teff", "st_mass",
        "st_rad", "pl_orbper", "pl_orbsmax",
        "pl_orbeccen", "st_met", "sy_snum", "sy_pnum"
    ]
    if c in planet_df.columns
]

ml_df = planet_df[ml_features + ["discoverymethod"]].copy()
ml_df = ml_df.dropna(subset=["discoverymethod"])

class_counts = ml_df["discoverymethod"].value_counts()
valid_classes = class_counts[class_counts >= 10].index
ml_df = ml_df[ml_df["discoverymethod"].isin(valid_classes)]

if len(ml_df) >= 20 and ml_df["discoverymethod"].nunique() >= 2:
    X = ml_df[ml_features].copy()
    y = ml_df["discoverymethod"].copy()
    X = X.dropna(axis=1, how="all")
    # Replace physically invalid non-positive measurements
    for col in [
        "pl_rade", "pl_eqt", "st_teff", "st_mass",
        "st_rad", "pl_orbper", "pl_orbsmax"
    ]:
        if col in X.columns:
            X.loc[X[col] <= 0, col] = np.nan

    label_encoder = LabelEncoder()
    y_encoded = label_encoder.fit_transform(y)

    # Stratify only if every class has enough samples
    can_stratify = (
        pd.Series(y_encoded).value_counts().min() >= 2
    )

    X_train, X_test, y_train, y_test = train_test_split(
        X,
        y_encoded,
        test_size=0.2,
        random_state=42,
        stratify=y_encoded if can_stratify else None
    )

    model = make_pipeline(
        SimpleImputer(strategy="median"),
        RandomForestClassifier(
            n_estimators=150,
            random_state=42,
            class_weight="balanced"
        )
    )

    model.fit(X_train, y_train)
    predictions = model.predict(X_test)

    print("\nDiscovery-method classification accuracy:",
          accuracy_score(y_test, predictions))

    print("\nClassification report:")
    print(
        classification_report(
            y_test,
            predictions,
            labels=np.arange(len(label_encoder.classes_)),
            target_names=label_encoder.classes_,
            zero_division=0
        )
    )

    ConfusionMatrixDisplay.from_predictions(
        y_test,
        predictions,
        labels=np.arange(len(label_encoder.classes_)),
        display_labels=label_encoder.classes_,
        xticks_rotation=45,
        cmap="Blues"
    )
    plt.title("Discovery Method Classification")
    plt.tight_layout()
    plt.show()

    # Feature importance from the trained random forest
    rf = model.named_steps["randomforestclassifier"]

    importance = pd.Series(
        rf.feature_importances_,
        index=X.columns
    ).sort_values(ascending=False)

    print("\nFeature importance:")
    print(importance)

    importance.head(10).sort_values().plot(
        kind="barh",
        title="Most Important Features for Classification"
    )
    plt.xlabel("Feature importance")
    plt.tight_layout()
    plt.show()

else:
    print("\nNot enough data for the selected classification task.")

# INTERACTIVE PLOTLY VISUALIZATION

try:
    import plotly.express as px

    interactive_df = planet_df[
        ["pl_rade", "pl_eqt", "discoverymethod"]
    ].dropna()

    interactive_df = interactive_df[
        (interactive_df["pl_rade"] > 0) &
        (interactive_df["pl_eqt"] > 0)
    ].head(3000)

    if len(interactive_df) > 0:
        fig = px.scatter(
            interactive_df,
            x="pl_rade",
            y="pl_eqt",
            color="discoverymethod",
            log_x=True,
            title="Interactive Exoplanet Analysis",
            labels={
                "pl_rade": "Planet Radius (Earth radii)",
                "pl_eqt": "Equilibrium Temperature (K)"
            }
        )
        fig.show()

except ImportError:
    print("Install plotly to enable interactive visualization.")

# SUMMARY
print("\n\nCAPSTONE ANALYSIS COMPLETED")
print("=" * 65)
print("Original rows:", len(df))
print("Rows after exact-duplicate removal:", len(analysis_df))
print("Unique planet names:", analysis_df["pl_name"].nunique())
print("Selected analysis columns:", len(planet_df.columns))
print("All five units have been covered.")
print("Output file: exoplanet_analysis_data.csv")
```
